**A. REPRODUCIBLE NORMALIZATION PROBLEM**

```python
import numpy as np
```

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5,5))
X
```

```python
x̄ = np.mean(X)
x̄
```

```python
σ = np.std(X)
σ
```

```python
X_normalized = (X - x̄)/σ
X_normalized
```

```python
x̄nmlzd = np.mean(X_normalized)
x̄nmlzd
```

```python
σnmlzd = np.std(X_normalized)
σnmlzd
```

```python
print("Array X", X)
```

```python
print("Normalized X Array", X_normalized)
```

```python
print("Normalized Mean = ", x̄nmlzd)
```

```python
print("Normalized Standard Deviation = ", σnmlzd)
```

```python
np.save('X_normalized.npy', X_normalized)
```

**B. CUBES DIVISIBLE BY 4 PROBLEM**

```python
integers = np.arange(1, 101)

cubed_integers = integers**3

C = cubed_integers.reshape(10, 10)
display(C)
```

```python
div_by_4 = C[C % 4 == 0]

print("Shape of C:", C.shape)
print("Array div_by_4:", div_by_4)
print("Number of selected elements:", div_by_4.size)
```

```python
np.save('div_by_4.npy', div_by_4)
```

**C. ABOVE-MEAN SQUARES PROBLEM**

```python
integers_36 = np.arange(1, 37)

squared_integers = integers_36**2

S = squared_integers.reshape(6, 6)
display(S)
```

```python
S_mean = np.mean(S)

above_mean = S[S > S_mean]

print("Array S:", S)
print("Mean of S (S_mean):", S_mean)
```

```python
print("Elements strictly greater than S_mean (above_mean):")
display(above_mean)
print("Number of selected elements:", above_mean.size)
```

```python
np.save('above_mean.npy', above_mean)
```
